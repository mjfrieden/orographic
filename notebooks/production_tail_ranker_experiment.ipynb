## Context & Methods

The objective is large after-friction option returns with explicit abstention. Outcomes are severe loss (≤ -50%), manageable loss, small win, and big win (≥ +50%). A four-class model estimates bucket probabilities; expected tail utility ranks contracts, while liquidity and Council gates remain binding.

### Key Assumptions

- One production lane and at most one live selection per scan.
- No probability-based position sizing.
- Frozen forward evidence is small and is suitable for active research, not sizing-grade claims.
- Disable active routing if 10 resolved live picks have negative cumulative after-friction return.
